# Exercise 1 — Speed Limit

A road has a speed limit of 120 km/h.

Complete the program so that if the car’s speed is greater than 120 km/h, its speed is changed to 120 km/h.

````python
distance = 1000
speed = 150
if speed > 120:
    speed = 120
time = distance / speed
print(f"Speed: {speed} km/h")
print(f"Time: {time:.2f} hours")
````

# Exercise 2 — Several Speeds
Now consider several possible speeds:
speeds = [80, 100, 120, 140]

Use a for loop to calculate the travel time for each speed over a distance of 1000 km.
The output should be similar to:

    Speed: 80 km/h - Time: 12.50 hours
    Speed: 100 km/h - Time: 10.00 hours
    Speed: 120 km/h - Time: 8.33 hours
    Speed: 140 km/h - Time: 7.14 hours

A Python list can be used to represent a vector. Consider the maximum speeds of four cars:
```
    names = ["Ferrari", "Porsche", "Tesla", "BMW"]
    speeds = [240, 220, 200, 250]
```

The position in one vector corresponds to the position in the other:
```
    names[0] -> Ferrari
    speeds[0] -> 240
```
We can therefore use a loop:
```
    for i in range(len(names)):
    print(names[i], speeds[i])
```

````python
speeds = [80, 100, 120, 140]
distance = 1000 #kms

for speed in speeds:
    average = distance / speed
    print(f"Speed: {speed} km/h - Time: {average} hours")


# Speed: 80 km/h - Time: 12.50 hours
# Speed: 100 km/h - Time: 10.00 hours
# Speed: 120 km/h - Time: 8.33 hours
# Speed: 140 km/h - Time: 7.14 hours
````

# Exercise 3 — Car Information
Modify the program to produce:
```
    Ferrari - 240 km/h
    Porsche - 220 km/h
    Tesla - 200 km/h
    BMW - 250 km/h
```
Then calculate the travel time for each car over 1000 km.

````python
names = ["Ferrari", "Porsche", "Tesla", "BMW"]
speeds = [240, 220, 200, 250]
distance = 1000
for i in range(len(names)):
    print(names[i],"-", speeds[i], "km/h")
    travel_time = distance/speeds[i]
    print("Travel time for 1.000 kms: ", travel_time, " hours.")
    print("")
````

# Exercise 4 — Display the Matrix
Modify the program so that each car’s speeds appear on one line.

The output should be:
```
    120 130 140
    100 110 120
    90 100 110
    130 140 150
```

````python
speeds = [
        [120, 130, 140],
        [100, 110, 120],
        [90, 100, 110],
        [130, 140, 150]
    ]

for row in speeds:
    row_values = ""
    for speed in row:
        row_values = row_values + " " + str(speed)
    row_values = row_values[1:]
    print(row_values)

````

# Exercise 5 — Average Speed
Calculate the average speed for each car.

The expected result is:
```
    Car 1: 130.00 km/h
    Car 2: 110.00 km/h
    Car 3: 100.00 km/h
    Car 4: 140.00 km/h
```

````python
speeds = [
        [120, 130, 140],
        [100, 110, 120],
        [90, 100, 110],
        [130, 140, 150]
    ]

def averageSpeed():
    print("Äverage speed")
    for i in range(len(speeds)):
        sum = 0
        for j in range(len(speeds[i])):
            sum += speeds[i][j]
        lenSpeed = len(speeds[i])
        print("Car ", i+1,": ", sum/lenSpeed," km/hour.")
averageSpeed()
````

# Exercise 8 — A Different Race

Create a race containing only:

• Audi — maximum speed: 225 km/h — energy rate: 0.07

• BMW — maximum speed: 230 km/h — energy rate: 0.07

Run the race for 1000 km.

````python
import random

class Car:
    def __init__(self, name, max_speed, energy_rate):
        self.name = name
        self.max_speed = max_speed
        self.energy_rate = energy_rate
    def run(self, distance):
        speed = random.randint(60, self.max_speed)
        time = distance / speed
        energy = distance * self.energy_rate
        return time, energy

class Race:
    def __init__(self, cars):
        self.cars = cars
        
    def run(self, distance):
        results = []
        for car in self.cars:
            time, energy = car.run(distance)
            results.append(
                (car.name, time, energy)
            )
        return results

cars = [
    Car("Audi", 225, 0.07),
    Car("BMW", 230, 0.07)
]
race = Race(cars)
results = race.run(1000)
for name, time, energy in results:
    print(
        f"{name}: "
        f"{time:.2f} hours - "
        f"{energy:.2f} units"
    )
````
